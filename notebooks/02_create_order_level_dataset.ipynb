```python
import pandas as pd
df = pd.read_csv("../data/gold/olist_final_consumption_dataset.csv")
```

```python
df.shape
```

```python
df['order_id'].nunique()
```

```python
aggregated_df = df.groupby('order_id').agg({
    'customer_id': 'first',
    'order_status': 'first',

    'order_purchase_timestamp': 'first',
    'order_delivered_customer_date': 'first',
    'order_estimated_delivery_date': 'first',

    'customer_state': 'first',
    'seller_state': lambda x: x.mode()[0],

    'product_category_name': 'first',
    'distance_bucket': 'first',

    'product_id': 'count',

    'product_weight_kg': 'sum',

    'seller_customer_distance_km': 'mean',

    'sla_breached': 'max'
}).reset_index()
```

```python
aggregated_df.shape
```

```python
aggregated_df['order_id'].nunique()
```

```python
aggregated_df.head()
```

```python
aggregated_df.info()
```

```python
aggregated_df = aggregated_df.rename(columns={
    'product_id': 'total_products'
})
```

```python
aggregated_df.to_csv(
    "../data/gold/base_order_dataset.csv",
    index=False
)
```
